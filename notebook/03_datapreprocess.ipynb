```python
# ===================== DATA PREPROCESSING =====================

import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load dataset
DATA_PATH = Path("../data/raw/WEC_Simulation_Data.csv")
df = pd.read_csv(DATA_PATH)

print(f"Original dataset shape: {df.shape}")

# 2. Define features and target
FEATURES = ["l1", "d1", "d2", "a1", "m1", "Qi"]
TARGET = "P_ext"

# 3. Keep only required columns
data = df[FEATURES + [TARGET]].copy()

# 4. Convert everything to numeric
for col in data.columns:
    data[col] = pd.to_numeric(data[col], errors="coerce")

# 5. Remove NaN and infinite values
data = data.replace([np.inf, -np.inf], np.nan)
data = data.dropna()

# 6. Remove physically invalid target values
data = data[data[TARGET] >= 0]

print(f"Cleaned dataset shape: {data.shape}")
print(f"Rows removed: {len(df) - len(data)}")

# 7. Separate features and target
X = data[FEATURES].values
y = data[TARGET].values.reshape(-1, 1)

# 8. Train-test split
# random_state ensures reproducibility
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# 9. Standardize features
# IMPORTANT: fit ONLY on training data
X_scaler = StandardScaler()

X_train_scaled = X_scaler.fit_transform(X_train)
X_test_scaled = X_scaler.transform(X_test)

# 10. Standardize target for ANN
y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(y_train)
y_test_scaled = y_scaler.transform(y_test)

# 11. Print final information
print("\nFeatures:", FEATURES)
print("Target:", TARGET)

print("\nTraining samples:", X_train_scaled.shape[0])
print("Testing samples :", X_test_scaled.shape[0])
print("Number of features:", X_train_scaled.shape[1])

print("\nScaled X_train mean:", X_train_scaled.mean(axis=0))
print("Scaled X_train std :", X_train_scaled.std(axis=0))

print("\nPreprocessing completed successfully.")
```
