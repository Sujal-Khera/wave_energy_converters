```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

DATA_PATH = Path("../data/raw/WEC_Simulation_Data.csv")

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["P_ext"], bins=100)

plt.xlabel("Extracted Power, P_ext")
plt.ylabel("Frequency")
plt.title("Distribution of Extracted Power")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.boxplot(df["P_ext"], vert=False)

plt.xlabel("Extracted Power, P_ext")
plt.title("P_ext Boxplot")

plt.tight_layout()
plt.show()
```

```python
analysis_cols = [
    "l1",
    "d1",
    "d2",
    "a1",
    "m1",
    "Qi",
    "P_ext"
]

corr = df[analysis_cols].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```
